1. NumPy: Sum of All Elements

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
total_sum = np.sum(arr)

print(f"Array: {arr}")
print(f"Sum of all elements: {total_sum}")
```

2. NumPy: 3x3 Matrix Operations

```python
import numpy as np

matrix = np.array([[12, 45, 7], 
                   [33, 89, 21], 
                   [54, 11, 90]])

# Row and Column sums
row_sum = np.sum(matrix, axis=1)
col_sum = np.sum(matrix, axis=0)

# 2nd Maximum Element
unique_sorted = np.sort(np.unique(matrix))
second_max = unique_sorted[-2]

print("3x3 Matrix:\n", matrix)
print("Sum of Rows:", row_sum)
print("Sum of Columns:", col_sum)
print("2nd Maximum Element:", second_max)
```

3. Matrix Multiplication ($n \times n$)

```python
import numpy as np

n = 2
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

result = np.matmul(A, B)

print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("Product (A*B):\n", result)
```

5. Pandas: Get First 3 Rows

```python
import pandas as pd
import numpy as np

exam_data = {
    'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily', 'Michael', 'Matthew', 'Laura', 'Kevin', 'Jonas'],
    'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
    'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    'qualify': ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']
}
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']

df = pd.DataFrame(exam_data, index=labels)

print("First three rows of the data frame:")
print(df.iloc[:3])
```

6. Pandas: Handle Missing Values

```python
# Using the DataFrame from the previous question
print("\nOriginal Score with NaNs:\n", df['score'])

# Replace NaN with 0
df_filled = df.fillna(0)

print("\nDataFrame after replacing missing values with 0:")
print(df_filled)
```

7. Matplotlib: Visual Forms

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 2]

plt.figure(figsize=(10, 4))

# 1. Line Plot
plt.subplot(1, 3, 1)
plt.plot(x, y, color='blue', marker='o')
plt.title("Line Plot")

# 2. Bar Chart
plt.subplot(1, 3, 2)
plt.bar(x, y, color='green')
plt.title("Bar Chart")

# 3. Scatter Plot
plt.subplot(1, 3, 3)
plt.scatter(x, y, color='red')
plt.title("Scatter Plot")

plt.tight_layout()
plt.show()
```
