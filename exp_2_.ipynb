1. Basic Arithmetic

```python
x = 9
y = 7
print(f"Addition: {x + y}")
print(f"Multiplication: {x * y}")
print(f"Division: {x / y}")
print(f"Subtraction: {x - y}")
```

2. Area of a Circle

```python
import math
r = float(input("Enter radius: "))
area = math.pi * (r ** 2)
print(f"Area of circle: {area:.2f}")
```

3, 4, & 5. Algebraic Expression (x+y)^2

```python
x = 4
y = 3
result = (x + y) * (x + y)
print(f"Result: {result}") 
```

6. Pythagoras Theorem

```python
import math
a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = math.sqrt(a**2 + b**2)
print(f"Hypotenuse: {c}")
```

7. Simple Interest

```python
p = float(input("Enter Principal: "))
r = float(input("Enter Rate: "))
t = float(input("Enter Time: "))
si = (p * r * t) / 100
print(f"Simple Interest: {si}")
```

8. Area of Triangle (Heron's Formula)

```python
import math
a = float(input("Side 1: "))
b = float(input("Side 2: "))
c = float(input("Side 3: "))
s = (a + b + c) / 2 
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
print(f"Area: {area}")
```

9. Convert Seconds to H:M:S

```python
total_sec = int(input("Enter seconds: "))
hours = total_sec // 3600
minutes = (total_sec % 3600) // 60
seconds = total_sec % 60
print(f"{hours} Hours, {minutes} Minutes, {seconds} Seconds")
```

10. Swap Two Numbers (Without Third Variable)

```python
a = int(input("Enter a: "))
b = int(input("Enter b: "))
a, b = b, a 
print(f"Swapped: a = {a}, b = {b}")
```

11. Sum of first $n$ Natural Numbers

```python
n = int(input("Enter n: "))
sum_n = (n * (n + 1)) // 2
print(f"Sum: {sum_n}")
```

12. Bitwise Truth Table

```python
print("A  B | A&B | A|B | A^B")
print("---------------")
for a, b in [(0,0), (0,1), (1,0), (1,1)]:
    print(f"{a}  {b} |  {a&b}  |  {a|b}  |  {a^b}")
```

13. Left and Right Shift

```python
num = int(input("Enter number: "))
shift = int(input("Shift by: "))
print(f"Left Shift ({num} << {shift}): {num << shift}")
print(f"Right Shift ({num} >> {shift}): {num >> shift}")
```

14. Membership Operator (Numbers)

```python
sequence = (10, 20, 56, 78, 89)
num = int(input("Enter number to check: "))
print(f"Is {num} in sequence? {num in sequence}")
```

15. Membership Operator (String)

```python
text = "Python Programming"
char = input("Enter character to check: ")
print(f"Is '{char}' in string? {char in text}")
```
