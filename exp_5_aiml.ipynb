```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

data = pd.read_csv("12th_marks_cgpa.csv")

print(data.head())

X = data[["12th_Marks"]]
y = data["CGPA"]

model = LinearRegression()

model.fit(X, y)

y_pred = model.predict(X)

print("\nCoefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
print("R2 Score:", r2_score(y, y_pred))
print("Mean Squared Error:", mean_squared_error(y, y_pred))

marks = 80
predicted_cgpa = model.predict([[marks]])

print("\nPredicted CGPA for", marks, "marks:", round(predicted_cgpa[0], 2))

print("Name: Fakre alam   Sap_id: 590020990")

plt.figure(figsize=(10, 6))

plt.scatter(X["12th_Marks"], y, label="Actual CGPA")
plt.plot(X["12th_Marks"], y_pred, label="Regression Line")

plt.xlabel("12th Marks (%)")
plt.ylabel("CGPA")
plt.title("Linear Regression: 12th Marks vs CGPA")
plt.figtext(0.99, 0.01, "Fakre Alam | SAP ID: 590020990",
            ha="right", va="bottom",
            fontsize=8, alpha=0.7)

plt.legend()
plt.grid()

plt.show()
```
