```python
class Toyota:
    def __init__(self, engine_l: float, weight_kg: float, tyre_in: float):
        """
        Initializes the Toyota vehicle with its specifications.
        """
        self.engine_l = engine_l
        self.weight_kg = weight_kg
        self.tyre_in = tyre_in

    def calculate_base_mileage(self) -> float:
        """
        Estimates the base mileage in km/L based on vehicle specs.
        """
        mileage = (28 
                   - (4 * self.engine_l) 
                   - (0.006 * (self.weight_kg - 1000)) 
                   - (0.7 * (self.tyre_in - 15)))
        
        return max(5.0, mileage)  # Clamp to minimum 5 km/L

    def get_mileage_report(self) -> dict:
        """
        Returns a dictionary containing base, city, and highway mileage.
        """
        base = self.calculate_base_mileage()
        return {
            "base": base,
            "city": 0.85 * base,
            "highway": 1.10 * base
        }

    def display_specs(self):
        """Prints a summary of the vehicle's efficiency."""
        stats = self.get_mileage_report()
        print(f"--- Toyota Vehicle Report ---")
        print(f"Specs: {self.engine_l}L engine, {self.weight_kg}kg, {self.tyre_in}in tyres")
        print(f"Base Mileage    : {stats['base']:.2f} km/L")
        print(f"City Mileage    : {stats['city']:.2f} km/L")
        print(f"Highway Mileage : {stats['highway']:.2f} km/L")
        print("-" * 28)


# Example usage:
if __name__ == "__main__":
    # Create an instance of the Toyota class
    my_car = Toyota(engine_l=1.5, weight_kg=1200, tyre_in=16)
    
    # Display the results
    my_car.display_specs()
```
