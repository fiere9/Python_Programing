```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

attendance_data = pd.read_csv("cgpa_attendance.csv")

X1 = attendance_data[["Attendance"]]
y1 = attendance_data["CGPA"]

model1 = LinearRegression()
model1.fit(X1, y1)

y1_pred = model1.predict(X1)

print("Example 1: CGPA vs Attendance")
print("Coefficient:", model1.coef_[0])
print("Intercept:", model1.intercept_)
print("R2 Score:", r2_score(y1, y1_pred))
print("Mean Squared Error:", mean_squared_error(y1, y1_pred))
print("Name: Fakre alam   Sap_id: 590020990")

plt.scatter(X1["Attendance"], y1)
plt.plot(X1["Attendance"], y1_pred)
plt.xlabel("Attendance (%)")
plt.ylabel("CGPA")
plt.title("Linear Regression: CGPA vs Attendance")
plt.figtext(0.99, 0.01, "Fakre Alam | SAP ID: 590020990",
            ha="right", va="bottom",
            fontsize=8, alpha=0.7)
plt.show()


maths_data = pd.read_csv("cgpa_maths.csv")

X2 = maths_data[["Maths_Marks"]]
y2 = maths_data["CGPA"]

model2 = LinearRegression()
model2.fit(X2, y2)

y2_pred = model2.predict(X2)

print("\nExample 2: CGPA vs Maths Marks")
print("Coefficient:", model2.coef_[0])
print("Intercept:", model2.intercept_)
print("R2 Score:", r2_score(y2, y2_pred))
print("Mean Squared Error:", mean_squared_error(y2, y2_pred))
print("Name: Fakre alam   Sap_id: 590020990")

plt.scatter(X2["Maths_Marks"], y2)
plt.plot(X2["Maths_Marks"], y2_pred)
plt.xlabel("Maths Marks")
plt.ylabel("CGPA")
plt.title("Linear Regression: CGPA vs Maths Marks")
plt.figtext(0.99, 0.01, "Fakre Alam | SAP ID: 590020990",
            ha="right", va="bottom",
            fontsize=8, alpha=0.7)

plt.show()
```
