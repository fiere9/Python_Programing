1. Handling names.txt

```python
# Setup: Creating the file
with open("name.txt", "w") as f:
    f.writelines(["Alice\n", "Oscar\n", "Umara\n", "Elizabeth\n", "Ben\n"])

# Processing the file
with open("name.txt", "r") as f:
    names = [line.strip() for line in f if line.strip()]

# a. Count number of names
count = len(names)

# b. Count names starting with a vowel
vowels = ('A', 'E', 'I', 'O', 'U')
vowel_names = [n for n in names if n.upper().startswith(vowels)]

# c. Longest name
longest = max(names, key=len)

print(f"Total names: {count}")
print(f"Names starting with vowels: {len(vowel_names)}")
print(f"Longest name: {longest}")
```

2. Integer Operations in Files

```python
# Setup: Creating the file
with open("numbers.txt", "w") as f:
    f.write("45\n120\n30\n250\n10\n")

with open("numbers.txt", "r") as f:
    nums = [int(line.strip()) for line in f]

# a. Max number
max_num = max(nums)

# b. Average
avg = sum(nums) / len(nums)

# c. Count > 100
gt_100 = len([n for n in nums if n > 100])

print(f"Max: {max_num}, Average: {avg}, Count > 100: {gt_100}")
```

4. Division with Exception Handling

```python
n = input("Enter number of test cases: ")
try:
    n = int(n)
    for _ in range(n):
        try:
            a, b = input("Enter a and b: ").split()
            print(int(a) // int(b))
        except ZeroDivisionError as e:
            print("Error Code:", e)
        except ValueError as e:
            print("Error Code:", e)
except ValueError:
    print("Invalid input for N")
```

5. File Handling with Multiple Exceptions

```python
def robust_file_reader(filename):
    try:
        with open(filename, 'r') as file:
            data = file.read()
            # Simulate a specific logic error
            if not data:
                raise EOFError("The file is empty!")
            print("File Content successfully read.")
            
    except FileNotFoundError:
        print("Error: The file you are looking for does not exist.")
    except PermissionError:
        print("Error: You do not have the required permissions to read this file.")
    except EOFError as e:
        print(f"Logic Error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    finally:
        print("File operation attempt completed.")

# Test it
robust_file_reader("non_existent_file.txt")
```
