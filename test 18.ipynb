## question 2

```python
def count_by_category(pairs):
    d = {}
    for cat, word in pairs:
        d.setdefault(cat, {})
        d[cat].setdefault(word, 0)
        d[cat][word] += 1
    return d

pairs = [("sports","win"), ("sports","win"), ("tech","ai")]
print(count_by_category(pairs))
```

### q 4

```python
def add_unique(x, seen=None):
    if seen is None:
        seen = set()
    
    if x not in seen:
        seen.add(x)
    
    return list(seen)


print(add_unique(1))  
print(add_unique(2))  
```
