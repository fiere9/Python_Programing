Q2) Armstrong Number

```python
num = int(input("Enter a number: "))
order = len(str(num))
temp = num
sum_val = 0

while temp > 0:
    digit = temp % 10
    sum_val += digit ** order
    temp //= 10

if num == sum_val:
    print(f"{num} is an Armstrong number")
else:
    print(f"{num} is not an Armstrong number")
```

Q3) Fibonacci Series

```python
n = int(input("Enter number of terms: "))
a, b = 0, 1
print("Fibonacci Series:", end=" ")
for _ in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

Q4) Prime Number Check

```python
num = int(input("Enter a number: "))
is_prime = True

if num <= 1:
    is_prime = False
for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
        is_prime = False
        break

print(f"{num} is {'a Prime' if is_prime else 'not a Prime'} number")
```

Q5) Palindrome Number

```python
num = int(input("Enter a number: "))
temp = num
reverse = 0

while temp > 0:
    reverse = (reverse * 10) + (temp % 10)
    temp //= 10

print(f"{num} is {'a Palindrome' if num == reverse else 'not a Palindrome'}")
```

Q6) Sum of Digits

```python
num = int(input("Enter a number: "))
sum_digits = 0
while num > 0:
    sum_digits += num % 10
    num //= 10
print(f"Sum of digits: {sum_digits}")
```

Q7) Numbers divisible by 5 or 7 (1-100)

```python
count = 0
print("Numbers:")
for i in range(1, 101):
    if i % 5 == 0 or i % 7 == 0:
        print(i, end=" ")
        count += 1
print(f"\nTotal Count: {count}")
```

Q8) Convert Lowercase to Uppercase

```python
text = input("Enter a string: ")
upper_text = ""
for char in text:
    if 'a' <= char <= 'z':
        upper_text += chr(ord(char) - 32)
    else:
        upper_text += char
print(f"Result: {upper_text}")
```

Q9) Prime Numbers Between 1 and 100

```python
print("Prime numbers between 1 and 100:")
for num in range(2, 101):
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            break
    else:
        print(num, end=" ")
```
