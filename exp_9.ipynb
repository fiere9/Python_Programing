 1. Basic Student Class

```python
class Student:
    def __init__(self, name, sap_id, marks):
        self.name = name
        self.sap_id = sap_id
        self.marks = marks  # List: [phy, chem, maths]

    def display(self):
        print(f"Name: {self.name} | SAP ID: {self.sap_id} | Marks: {self.marks}")

students = []
for i in range(3):
    print(f"Enter details for Student {i+1}:")
    name = input("Name: ")
    sap = input("SAP ID: ")
    m_phy = float(input("Physics Marks: "))
    m_chem = float(input("Chemistry Marks: "))
    m_math = float(input("Maths Marks: "))
    students.append(Student(name, sap, [m_phy, m_chem, m_math]))

print("\n--- Student Details ---")
for s in students:
    s.display()
```

2. Advanced Student Class (Constructor & Logic)

```python
class Student:
    def __init__(self, name, sap_id, marks):
        self.name = name
        self.sap_id = sap_id
        self.marks = marks # [phy, chem, maths]

    def display(self):
        print(f"{self.name}\t{self.sap_id}\t{self.marks}\t{self.marks_percentage():.2f}%\t{self.result()}")

    def marks_percentage(self):
        return sum(self.marks) / 3

    def result(self):
        # Check if every subject score is > 40
        return "Pass" if all(m > 40 for m in self.marks) else "Fail"

def class_average(student_list):
    total_avg = sum(s.marks_percentage() for s in student_list)
    return total_avg / len(student_list)

# Implementation
n = int(input("Enter number of students: "))
student_objects = []
for i in range(n):
    name = input("Name: ")
    sap = input("SAP ID: ")
    m = [float(x) for x in input("Enter Physics, Chem, Maths marks: ").split()]
    student_objects.append(Student(name, sap, m))

print("\nName\tSAP ID\tMarks\t\tPercentage\tResult")
for s in student_objects:
    s.display()

print(f"\nOverall Class Average: {class_average(student_objects):.2f}%")
```

3. Types of Inheritance

```python
# Single Inheritance
class Parent:
    def func1(self): print("Parent Class")

class Child(Parent):
    def func2(self): print("Child Class")

# Multilevel Inheritance
class GrandChild(Child):
    def func3(self): print("GrandChild Class")

# Multiple Inheritance
class Father:
    def f_name(self): print("Father")

class Mother:
    def m_name(self): print("Mother")

class Son(Father, Mother):
    def s_name(self): print("Son")

print("--- Inheritance Demo ---")
obj = Son()
obj.f_name()
obj.m_name()
```

4. Method Overriding

```python
class Animal:
    def sound(self):
        print("Animals make sounds")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")

# Implementation
a = Animal()
d = Dog()
c = Cat()

a.sound() # Base class
d.sound() # Overridden
c.sound() # Overridden
```

5. Operator Overloading (Point Addition)

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading the + operator
    def __add__(self, other):
        new_x = self.x + other.x
        new_y = self.y + other.y
        return Point(new_x, new_y)

    def __str__(self):
        return f"Point(x={self.x}, y={self.y})"

# Implementation
p1 = Point(10, 20)
p2 = Point(12, 15)
p3 = p1 + p2

print(f"P1: {p1}")
print(f"P2: {p2}")
print(f"P3 (P1 + P2): {p3}")
```
