2. Print Strings in Specific Formats

```python
# a) Hello Everyone !!!
print("Hello Everyone !!!")

# b) Hello (new line) World
print("Hello\nWorld")

# c) Using triple quotes for multi-line formatting
print("""Hello
World""")

# d) Using escape characters for quotes and backslashes
print("' Rohit' s date of birth is 12\\05\\1999'")
```

4. Print Different Data Types

```python
integer_val = 25
float_val = 98.6
string_val = "Python"
boolean_val = True

print(f"Integer: {integer_val}")
print(f"Float: {float_val}")
print(f"String: {string_val}")
print(f"Boolean: {boolean_val}")
```

6. Formatting with Parentheses

```python
first_name = "George"
last_name = "Washington"
nickname = "woody"

print(f"{first_name} ({nickname}) {last_name}")
```
