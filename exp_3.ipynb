1. Divisible by 3 and 5

```python
num = int(input("Enter a number: "))
if num % 3 == 0 and num % 5 == 0:
    print("Divisible by both 3 and 5")
else:
    print("Not divisible by both")
```

2. Multiple of 5

```python
num = int(input("Enter a number: "))
if num % 5 == 0:
    print("Multiple of 5")
else:
    print("Not a multiple of 5")
```

3. Greatest of Two Numbers

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
if a > b:
    print(f"{a} is greater")
elif b > a:
    print(f"{b} is greater")
else:
    print("Numbers are equal")
```

4. Greatest of Three Numbers

```python
a, b, c = map(int, input("Enter three numbers separated by space: ").split())
if a > b and a > c:
    print(f"{a} is greatest")
elif b > c:
    print(f"{b} is greatest")
else:
    print(f"{c} is greatest")
```

5. Quadratic Equation Roots

```python
import math
a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

d = b**2 - 4*a*c

if d >= 0:
    root1 = (-b + math.sqrt(d)) / (2*a)
    root2 = (-b - math.sqrt(d)) / (2*a)
    print(f"Real Roots: {root1} and {root2}")
else:
    real_part = -b / (2*a)
    imag_part = math.sqrt(-d) / (2*a)
    print(f"Imaginary Roots: {real_part} + {imag_part}i and {real_part} - {imag_part}i")
```

6. Leap Year Check

```python
year = int(input("Enter year: "))
if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

7. Next Date Calculation

```python
d = int(input("day="))
m = int(input("month="))
y = int(input("year="))

# Days in each month
if m in [1, 3, 5, 7, 8, 10, 12]:
    max_days = 31
elif m in [4, 6, 9, 11]:
    max_days = 30
elif (y % 400 == 0) or (y % 4 == 0 and y % 100 != 0):
    max_days = 29
else:
    max_days = 28

if d < max_days:
    d += 1
else:
    d = 1
    if m == 12:
        m = 1
        y += 1
    else:
        m += 1

print(f"Next Date: day={d} month={m} year={y}")
```

8. Gradesheet Program

```python
name = input("Enter Name: ")
roll = input("Enter Roll Number: ")
sap = input("Enter SAPID: ")
sem = input("Enter Sem: ")
course = input("Enter Course: ")

subjects = ["PDS", "Python", "Chemistry", "English", "Physics"]
marks = []
for sub in subjects:
    marks.append(float(input(f"Enter marks for {sub}: ")))

total = sum(marks)
percentage = total / 5
cgpa = percentage / 10

if 9.1 <= cgpa <= 10: grade = "O"
elif 8.1 <= cgpa <= 9: grade = "A+"
elif 7.1 <= cgpa <= 8: grade = "A"
elif 6.1 <= cgpa <= 7: grade = "B+"
elif 5.1 <= cgpa <= 6: grade = "B"
elif 3.5 <= cgpa <= 5: grade = "C+"
else: grade = "F"

print("\n" + "="*30)
print(f"Name: {name}")
print(f"Roll Number: {roll} SAPID: {sap}")
print(f"Sem: {sem} Course: {course}")
print("-" * 30)
for i in range(5):
    print(f"{subjects[i]}: {marks[i]}")
print("-" * 30)
print(f"Percentage: {percentage}%")
print(f"CGPA: {cgpa:.1f}")
print(f"Grade: {grade}")
```
