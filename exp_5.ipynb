#### 1. Count Capital Letters

```python
text = input("Enter a string: ")
capitals = [char for char in text if char.isupper()]
print(f"Capital letters: {''.join(capitals)}")
print(f"Count: {len(capitals)}")
```

#### 2. Count Total Vowels

```python
text = input("Enter a string: ").lower()
vowels = "aeiou"
count = sum(1 for char in text if char in vowels)
print(f"Total number of vowels: {count}")
```

#### 3. Print Words in Separate Lines

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
for word in words:
    print(word)
```

#### 4. Count Substring Occurrences (Overlapping)

```python
string = input("Enter string: ")
sub = input("Enter substring: ")
count = 0
for i in range(len(string) - len(sub) + 1):
    if string[i : i + len(sub)] == sub:
        count += 1
print(count)
```

#### 5. Case-Insensitive Alphabet Count

```python
string = input("Enter string: ").upper()
checked = ""
for char in string:
    if char.isalpha() and char not in checked:
        print(f"{string.count(char)}{char}")
        checked += char
```

#### 6. Count Unique Words Using Sets

```python
sentence = input("Enter a sentence: ").lower()
words = sentence.split()
unique_words = set(words)
print(f"Number of unique words: {len(unique_words)}")
```

#### 7. Fruit Set Operations

```python
n = int(input("Enter number of fruits (n): "))
s1 = {input(f"Set 1 - Fruit {i+1}: ") for i in range(n)}
s2 = {input(f"Set 2 - Fruit {i+1}: ") for i in range(n)}

print(f"a. Fruits in both: {s1 & s2}")
print(f"b. Only in s1: {s1 - s2}")
print(f"c. Total unique fruits across both: {len(s1 | s2)}")
```

#### 8.Set Operations Comparison

```python
S1 = {"Red", "yellow", "orange", "blue"}
S2 = {"violet", "blue", "purple"}
union_res = S1 | S2
inter_res = S1 & S2
diff_res = S1 - S2
sym_diff = S1 ^ S2
print(f"Union: {union_res}")
print(f"Intersection: {inter_res}")
print(f"Difference (S1-S2): {diff_res}")
print(f"Symmetric Difference: {sym_diff}")
```
