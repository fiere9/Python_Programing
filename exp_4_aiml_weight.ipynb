```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv("weight_data.csv")

X = data[["Height", "Age", "Calories"]].values
y = data["Weight"].values

X_mean = X.mean(axis=0)
X_std = X.std(axis=0)

X_normalized = (X - X_mean) / X_std

m = len(y)

X_bias = np.c_[np.ones(m), X_normalized]

theta = np.zeros(4)

learning_rate = 0.01
iterations = 1000

cost_history = []

for i in range(iterations):

    y_pred = X_bias.dot(theta)

    error = y_pred - y

    gradient = (1 / m) * X_bias.T.dot(error)

    theta = theta - learning_rate * gradient

    cost = (1 / (2 * m)) * np.sum(error ** 2)

    cost_history.append(cost)


# Final predictions
predicted_data = X_bias.dot(theta)

print("Final Weights:")
print("Bias:", theta[0])
print("Height Weight:", theta[1])
print("Age Weight:", theta[2])
print("Calories Weight:", theta[3])
print("Name: Fakre alam   Sap_id: 590020990")

# Cost vs Iterations
plt.figure(figsize=(10, 6))

plt.plot(range(iterations), cost_history)

plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Gradient Descent - Cost vs Iterations")
plt.grid()
plt.figtext(0.99, 0.01, "Fakre Alam | SAP ID: 590020990",
            ha="right", va="bottom",
            fontsize=8, alpha=0.7)
plt.show()


# Original Data vs Predicted Data
plt.figure(figsize=(10, 6))

sample_index = np.arange(len(y))

plt.scatter(sample_index, y, label="Original Data")

plt.scatter(
    sample_index,
    predicted_data,
    marker="x",
    label="Predicted Data"
)

plt.xlabel("Sample Index")
plt.ylabel("Weight")
plt.title("Multiple Linear Regression - Original vs Predicted Weight")
plt.figtext(0.99, 0.01, "Fakre Alam | SAP ID: 590020990",
            ha="right", va="bottom",
            fontsize=8, alpha=0.7)

plt.legend()
plt.grid()

plt.show()
```
