#### Frequency Count (0-3 Range)

```python
n = int(input("Enter number of elements: "))
elements = []
for _ in range(n):
    val = int(input("Enter value (0-3): "))
    elements.append(val)

for i in range(4):
    print(f"{i} occurred {elements.count(i)} times")
```

#### Average of a Tuple

```python
n = int(input("Enter number of values: "))
values = tuple(float(input(f"Enter value {i+1}: ")) for i in range(n))

if n > 0:
    avg = sum(values) / n
    print(f"The average is: {avg}")
```

#### Finding the Runner-Up Score

```python
n = int(input("N = "))
scores = list(map(int, input("Scores = ").split()))

unique_scores = list(set(scores))
unique_scores.sort(reverse=True)

if len(unique_scores) > 1:
    print(f"Runner-up score: {unique_scores[1]}")
else:
    print("No runner-up available.")
```

#### Name and City Dictionary

```python
n = int(input("Enter number of persons: "))
records = {}

for _ in range(n):
    name = input("Enter Name: ")
    city = input("Enter City: ")
    records[name] = city

print("\na. Names:", list(records.keys()))
print("b. Cities:", list(records.values()))
print("c. Full Records:", records)


city_counts = {}
for city in records.values():
    city_counts[city] = city_counts.get(city, 0) + 1
print("d. Student count per city:", city_counts)
```

#### Movie Database System

```python
n = int(input("Enter number of movies: "))
movies = []

for i in range(n):
    print(f"\nMovie {i+1} Details:")
    movie = {
        "name": input("Name: "),
        "year": int(input("Year: ")),
        "director": input("Director: "),
        "cost": float(input("Production Cost: ")),
        "earnings": float(input("Earnings: "))
    }
    movies.append(movie)

print("\n--- a. All Movie Details ---")
for m in movies:
    print(m)

print("\n--- b. Released before 2015 ---")
for m in movies:
    if m['year'] < 2015:
        print(m['name'])

print("\n--- c. Profitable Movies ---")
for m in movies:
    if m['earnings'] > m['cost']:
        print(f"{m['name']} (Profit: {m['earnings'] - m['cost']})")
```
