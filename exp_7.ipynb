1. Max and Min (Manual Logic)

```python
def find_max_min(numbers):
    if not numbers:
        return None, None
    
    max_val = numbers[0]
    min_val = numbers[0]
    
    for num in numbers:
        if num > max_val:
            max_val = num
        if num < min_val:
            min_val = num
            
    return max_val, min_val
print("Result:", find_max_min([10, 6, 8, 90, 12, 56]))
```

2. Sum of Cubes

```python
def sum_of_cubes(n):
    total = 0
    for i in range(1, n):
        total += i**3
    return total
print("Result:", sum_of_cubes(5))
```

3. Print 1 to $n$ (Recursion)

```python
def print_up_to_n(n):
    if n > 0:
        print_up_to_n(n - 1)
        print(n)
print_up_to_n(5)
print()        
```

4. Fibonacci Series (Recursion)

```python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)

def print_fib_series(terms):
    for i in range(terms):
        print(fibonacci(i), end=" ")
for i in range(5):
    print(fibonacci(i), end=" ")
print()        
```

5. Lambda: Volume of a Cone

```python
import math
volume_cone = lambda r, h: (1/3) * math.pi * (r**2) * h
print(f"Volume: {volume_cone(3, 5):.2f}")
```

6. Lambda: Max and Min Tuple

```python
max_min_lambda = lambda lst: (max(lst), min(lst))
sample_input = [10, 6, 8, 90, 12, 56]
print(max_min_lambda(sample_input)) 
sample_list = [10, 6, 8, 90, 12, 56]
print("Output:", max_min_lambda(sample_list))
```

```python
# 7.a. Keyword Argument
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type} named {pet_name}.")
describe_pet(pet_name="Whiskers", animal_type="Cat")
# b. Default Argument
def power(base, exponent=2):
    return base ** exponent
result_default = power(5)         
result_provided = power(5, 3)      
print(f"5 squared (default): {result_default}")
print(f"5 cubed (provided): {result_provided}")
# c. Variable Length Argument (*args)
def sum_everything(*args):
    return sum(args)
print(f"Sum of (10, 20): {sum_everything(10, 20)}")
print(f"Sum of (1, 2, 3, 4, 5): {sum_everything(1, 2, 3, 4, 5)}")
```
