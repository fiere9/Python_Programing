```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

np.random.seed(42)

X = np.random.rand(100, 1) * 10
y = 3 * X.flatten() + 5 + np.random.randn(100) * 2

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)

new_value = np.array([[6]])
prediction = model.predict(new_value)

print("Prediction for X = 6:", prediction[0])
print("Name: Fakre alam   Sap_id: 590020990")

plt.scatter(X, y, label="Random Data")
plt.plot(X, y_pred, label="Regression Line")
plt.scatter(new_value, prediction, label="Predicted Value")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression on Random Data")
plt.figtext(0.99, 0.01, "Fakre Alam | SAP ID: 590020990",
            ha="right", va="bottom",
            fontsize=8, alpha=0.7)

plt.legend()
plt.show()
```
