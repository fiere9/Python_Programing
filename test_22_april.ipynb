Q6. Magic Methods / Operator Overloading

```python
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def __str__(self):
        return f"product:{self.name} | price: ${self.price}"
    def __add__(self,other):
      if isinstance(other,product):  
         return self.price + other.price
      return NotImplemented
    def __eq__(self,other):
       if isinstance(other,product):
           return self.name.lower() == other.name.lower()
       return False

p1 = product("Keyboard",50)
p2 = product("Mouse",45)
p3 = product("Keyboard",50)

print(p1)
print(p1 + p2)
print(p1 == p2)
print(p1 == p3)
    
```

Q4. Public and Private Data Members

```python
class BankAccount:
    def __init__(self, name, initial_balance):
        # Public variable: accessible from anywhere
        self.name = name
        # Private variable: accessible only within this class
        self.__balance = initial_balance

    def deposit(self, amount):
        """Increases the balance by the specified amount."""
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ${amount}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        """Reduces balance if funds are sufficient; otherwise, prints error."""
        if amount > self.__balance:
            print("Insufficient balance")
        elif amount <= 0:
            print("Withdrawal amount must be positive.")
        else:
            self.__balance -= amount
            print(f"Withdrew: ${amount}")

    def show_balance(self):
        """Prints the current private balance."""
        print(f"Account Holder: {self.name}")
        print(f"Current Balance: ${self.__balance}")

# --- Testing the Methods ---

# 1. Create an object
account = BankAccount("John Doe", 1000)

# 2. Test deposit
account.deposit(500)

# 3. Test withdraw (Sufficient balance)
account.withdraw(200)

# 4. Test withdraw (Insufficient balance)
account.withdraw(2000)

# 5. Show final balance
account.show_balance()
```
